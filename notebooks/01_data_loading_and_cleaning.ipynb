#### Employment length in years (person_emp_length)

Empty fields are set to 0

```python
df["emp_length_missing"] = df["person_emp_length"].isnull().astype(int)
df["person_emp_length"] = df["person_emp_length"].fillna(df["person_emp_length"].median())
```

```python
df["person_emp_length"].isnull().sum()
```

#### Interesst rate (loan_int_rate)

```python
df[df["loan_int_rate"].isnull()]["loan_status"].value_counts()
```

#### Loan

Loan interest rate: "null" are removed

```python
print(df["loan_int_rate"].isnull().sum())

# Drop rows with missing interest rate
df = df.dropna(subset=["loan_int_rate"])

# Verify
print(df["loan_int_rate"].isnull().sum())
print(df.shape)
```
